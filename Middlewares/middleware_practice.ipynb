```python
class MySimpleMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        response=self.get_response(request)
        return response
```

```python
class MyLoggingMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        print(f'Incoming request from path: {request.path}')

        #calling the next middleware or view
        response=self.get_response(request)

        print(f"Outgoing response statu: {response.status_code}")

        return response
```

```python
#modifying the request and the response with custome middlwares
class MyCustomMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        #modifying request
        request.my_customer_attr="Hello, Middleware"

        response=self.get_response(request)

        response['X-Custom-Header']='Middleware was here'

        return response
```

```python
#creating an authentication checker middleware
class TokenAuthMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        token=request.headers.get('X-Auth-tken',None)
        if not token:
            print("Token not provided")

        response=self.get_response(request)
        return response
```

```python
#creating a custom request timer middleware
import time
class RequestTimerMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        start_time=time.time()

        response=self.get_response(request)
        duration=time.time()-start_time

        print(f"Request to {request.path} took {duration:.2f} seconds")
        return response
```

```python
import time

class FullFeatureMiddleware:
    def __init__(self,get_response):
        self.get_response=get_response

    def __call__(self,request):
        start_time=time.time()

        print(f'[Request] Path: {request.path}')

        token=request.headers.get('X-Auth-Token',None)
        if not token:
            print('[Auth] No token provided !')

        response=self.get_response(request)
        response['X-Frame-Options']='DENY'
        response['X-Content-Type-Options']='nosniff'

        duration = time.time() - start_time
        print(f"[Response] Path: {request.path} took {duration:.2f} sec, Status: {response.status_code}")

        return response
```

```python
class BlockPathMiddleware:
    def __init__(self, get_response):
        self.get_response = get_response

    def __call__(self, request):
        blocked = ["/blocked"]
        
        if request.path in blocked:
            print("Blocked path accessed!")
            return "Fake response: blocked"  # Syntax placeholder for HttpResponse
        
        response = self.get_response(request)
        return response
```
