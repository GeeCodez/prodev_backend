```python
class StudentIterator:
    def __init__(self,students):
        self.students=students
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.students):
            raise StopIteration
        student=self.students[self.index]
        self.index+=1
        return student
    
students=["Hello","Hi","Wi","why","they"]
for i in StudentIterator(students):
    print(i)
```

```python
#bookiteratorclass
class BookIterator:
    def __init__(self,books):
        self.books=books
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.books):
            raise StopIteration
        next_book=self.books[self.index]
        self.index+=1
        return next_book
    
books=("Today","Monday","Lastday","thisday")
for book in BookIterator(books):
    print(book)
```

```python
import random

class ShipmentIterator:
    def __init__(self, shipments):
        self.shipments = shipments
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.shipments):
            raise StopIteration
        shipment = self.shipments[self.index]
        self.index += 1
        return shipment


shipments = [
    ("Milk", "Kumasi"),
    ("Apples", "Accra"),
    ("Juice", "Takoradi"),
    ("Groceries", "Sunyani")
]

for product, destination in ShipmentIterator(shipments):
    order_id = f"{product[:3].upper()}{random.randint(10000,99999)}"
    print(f"Processing order_id {order_id} → Destination: {destination}")
```

```python
#creating a custome iterator for managing transactions
import time
class TransactionIterator:
    def __init__(self,transactions):
        self.transactions=transactions
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.transactions):
            raise StopIteration
        next_transaction=self.transactions[self.index]
        self.index+=1
        return next_transaction
    
transactions = (
    ("John Doe", 250.00, "deposit"),
    ("Mary Mensah", 120.50, "withdrawal"),
    ("Kojo Asare", 500.00, "deposit"),
    ("Ama Boateng", 75.25, "withdrawal"),
    ("Yaw Owusu", 1000.00, "deposit")
)

for name, amount, tp in TransactionIterator(transactions):
    print(f"Processing a {tp} of {amount} for {name}")
```

```python
import time
#creating a sensor Iterator
class SensorIterator:
    def __init__(self,readings):
        self.readings=readings
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=len(self.readings):
            raise StopIteration
        reading=self.readings[self.index]
        self.index+=1
        return reading
    
sensor_readings = (
    ("Accra", 29.4, "2025-11-10 08:00:00"),
    ("Kumasi", 27.8, "2025-11-10 08:05:00"),
    ("Takoradi", 30.1, "2025-11-10 08:10:00"),
    ("Sunyani", 26.9, "2025-11-10 08:15:00"),
    ("Tamale", 33.2, "2025-11-10 08:20:00"),
    ("Cape Coast", 28.3, "2025-11-10 08:25:00"),
    ("Ho", 27.5, "2025-11-10 08:30:00"),
    ("Bolgatanga", 35.0, "2025-11-10 08:35:00")
)

for city,temperature,date in SensorIterator(sensor_readings):
    print(f"Sensor reading from {city}: {temperature} at {date}")
    time.sleep(1)
```

```python
import time
import sys

# Set countdown in seconds (e.g., 1 minute = 60 seconds)
total_seconds = 60

for remaining in range(total_seconds, 0, -1):
    mins, secs = divmod(remaining, 60)  # get minutes and seconds
    current_time = time.strftime("%I:%M:%S %p", time.localtime())
    
    # Progress bar
    progress_length = 20
    filled_length = progress_length * (total_seconds - remaining) // total_seconds
    bar = "#" * filled_length + "-" * (progress_length - filled_length)
    
    sys.stdout.write(f"\r[{bar}] {mins:02d}:{secs:02d} - Current time: {current_time}")
    sys.stdout.flush()
    time.sleep(1)

# Blast off
current_time = time.strftime("%I:%M:%S %p", time.localtime())
print(f"\nBlast off! - Current time: {current_time}")
```
